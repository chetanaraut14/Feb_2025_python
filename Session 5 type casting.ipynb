##  Type casting

- Convert one data type to another data type

- we learned 4 data type

    - integer : int

    - float  : float

    - boolean : bool

    - string : str

**int to other data types**

```python
number=100
type(number)
```

```python
float_number=float(number)
type(float_number)
```

```python
bool_number=bool(number)
type(bool_number)
```

```python
str_number=str(number)
type(str_number)
```

```python
float(100), bool(100), str(100)
```

```python
print(float(100))
print(bool(100))
print(str(100))
```

```python
float(-100),bool(-100),str(-100)
```
