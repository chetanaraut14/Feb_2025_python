**OCTA NUMBER SYSTEM**

- It is denoted with **0o**  or **0O**

- Octa means  8

- We have generally 0 to 9 digits available

- octa means only  digit allowed : 0 1 2 3 4 5 6 7

- ex: 0o567 0o111 0o001 0o672

- not valid : 0o768 0o169

```python
0o123
```

```python
0o567
```

**HEXA NUMBER SYSTEM**

- It is denoted with **0x**  or  **0X**

- hexa means 16

- We have generally 0 to 9 digits available

- octa means only digits allowed : 0 to 9 A to F

- A:10   B:11  C:12  D:13  E:14  F:15

- ex :0xabc  0xA1

- not valid : 0xefg
